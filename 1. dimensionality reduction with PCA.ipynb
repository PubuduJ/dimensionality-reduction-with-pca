```python
# Import the Pandas library for loading, manipulating, and analysing structured datasets.
# It is commonly used to work with data in rows and columns (DataFrames).
import pandas as pd

# Import the NumPy library for performing numerical and mathematical operations,
# especially when working with arrays and numerical data.
import numpy as np

# Import the Pyplot module from Matplotlib for creating data visualizations,
# such as line charts, bar charts, scatter plots, and histograms.
import matplotlib.pyplot as plt
```

```python
# -----------------------------------------------------------------------------
# Load the Breast Cancer Dataset
# ----------------------------------------------------------------------------- 

# Load the Breast Cancer dataset from the CSV file into a Pandas DataFrame.
# The "target" column is removed because we only need the feature values for the upcoming analysis.
df = pd.read_csv("assets/breast_cancer_dataset.csv").drop(columns=["target"])

# Display the first 5 rows to get a quick overview of the feature data.
print("First 5 rows of the dataset:")
print(df.head())

# Display the dimensions of the dataset.
# The shape shows the number of samples (rows) and features (columns).
print(f"Original Dataset Shape: {df.shape}")
```

```python

# -----------------------------------------------------------------------------
# Standardize the Feature Values
# -----------------------------------------------------------------------------

# Import StandardScaler from Scikit-learn to standardize the feature values.
# Standardization transforms the data so that each feature has a mean close to 0 and a standard deviation close to 1.
from sklearn.preprocessing import StandardScaler

# Create a StandardScaler object that will be used to standardize the data.
scaler = StandardScaler()

# Fit the scaler to the dataset and transform each feature into its standardized "Z-score" value.
scaled_data = scaler.fit_transform(df)

# Display a message confirming that the feature values have been standardized.
print("Data is now scaled! Mean is approximately 0 and standard deviation is 1.")

# Display the shape of the scaled dataset.
# The shape shows the number of samples (rows) and features (columns).
print(scaled_data.shape)

# Display the first 5 rows of the standardized data.
print(scaled_data[:5])
```

```python
# -----------------------------------------------------------------------------
# Apply Principal Component Analysis (PCA)
# -----------------------------------------------------------------------------

# Import PCA from Scikit-learn.
# PCA is a dimensionality reduction technique that transforms the original features into a smaller set of new features called principal components.
from sklearn.decomposition import PCA

# Create a PCA object with all 30 components (Feature count == component count).
# At this stage, we keep all components so that we can examine how much variance is explained by each component.
# if PCA(n_components=0.9), set to a decimal value like 0.9,
#   then Scikit-learn automatically selects the number of components needed to retain approximately 90% of the variance.
pca = PCA(n_components=30)

# Fit PCA to the standardized data and transform the original features into principal components.
pca_data = pca.fit_transform(scaled_data)

# Display the shape of the PCA-transformed dataset.
# Since all 30 components are retained, the number of features remains 30.
print(f"PCA dataset shape: {pca_data.shape}")

# Display the first 5 rows of the PCA-transformed data.
print(pca_data[:5])
```

```python
# -----------------------------------------------------------------------------
# Visualize the Cumulative Explained Variance
# -----------------------------------------------------------------------------

# Calculate the cumulative explained variance for the principal components.
# explained_variance_ratio_ contains the proportion of variance explained by each component, while cumsum() adds these values progressively.
cumulative_variance = np.cumsum(pca.explained_variance_ratio_)

# Create a figure for the cumulative variance plot.
plt.figure(figsize=(8, 5))

# Plot the cumulative explained variance against the number of components.
# This helps us identify how many components are needed to retain a desired amount of information from the original dataset.
plt.plot(
    range(1, 31),
    cumulative_variance,
    marker="o",
    linestyle="--"
)

# Add a title and labels to make the plot easier to understand.
plt.title("Cumulative Explained Variance by Principal Components")
plt.xlabel("Number of Components (k)")
plt.ylabel("Cumulative Explained Variance")

# Add a horizontal reference line at 90% explained variance.
# This helps identify the number of components required to retain approximately 90% of the information in the original dataset.
plt.axhline(
    y=0.90,
    color="r",
    linestyle="-"
)

# Add a label to identify the 90% variance threshold on the plot.
plt.text(
    5,
    0.85,
    "90% Variance Threshold",
    color="red"
)

# Display grid lines to make it easier to read the values from the plot.
plt.grid(True)

# Display the completed plot.
plt.show()
```
